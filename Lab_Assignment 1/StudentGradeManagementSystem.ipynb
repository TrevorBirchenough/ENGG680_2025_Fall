```python
# Classes and Function (3 marks)

# Step 1: Create a class BankAccount.
# Add class attribute bankname="Canada Bank"
# Add 2 instance attributes: account_holder (string) & balance (float, default value 0.0)

# Step 2: Add methods
# 1. deposit(amount) -> Add amount to the balance
# 2. withdraw(amount) -> Subtract the amount if balance is sufficient, else print an error message
# 3. get_balance() -> Print Account holder's name and current balance
# 4. get_bankname() -> Print the class attribute (staticmethod)
# Note: Use f-strings to display the amount and updated balance whenever you call deposit() or withdraw()


# Step 3: Test
# 1. Print the Bankname
# 2. Create an object of BankAccount with your name and an initial balance of 1000
# 3. Deposit 500
# 4. Withdraw 200
# 5. Print the final balance
```

```python
class BankAccount:
    #class attribute
    bankname="Canada Bank"
    #instance attributes
    def __init__(self, account_holder, balance=0.0):
        self.account_holder = account_holder
        self.balance = balance

    #deposit method
    def deposit(self, amount):
        self.balance += amount
        print(f"{self.account_holder} deposited ${amount}. New balance: ${self.balance}")

    #withdraw method
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"{self.account_holder} withdrew ${amount}. Remaining balance: ${self.balance}")
        else:
            print(f"Insufficient funds! {self.account_holder} tried to withdraw ${amount} but only has ${self.balance}")

    #get balance
    def get_balance(self):
        print(f"{self.account_holder}'s current balance: ${self.balance}")

    #static method to get bank name
    @staticmethod
    def get_bankname():
        print(f"Bank Name: {BankAccount.bankname}")

#print the Bankname
BankAccount.get_bankname()

#Create an account
account1= BankAccount("Jason", 1000)

#Deposit 500
account1.deposit(500)

#Withdraw 200
account1.withdraw(200)

#Print the final balance
account1.get_balance()
```

## Part 6: Linear LSE
### (35 marks)

### Curve Fitting

curvefitting.png

### Observations: values of a function f(a) in range of -π to π, contaminated with random noise

### Unknown States: coefficients of polynomial functions for curve fitting

```python
# step 1
# import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# step 2
# Load data.xlsx file, which contains the observations (with noise), variance of observations and true observations
# hint: use pandas library to parse excel file and use numpy library to convert it to array
# (1 mark)
df_data = pd.read_excel('data.xlsx')
arr_data = df_data.values

#store the noisy observations, variance of observations and true observations into numpy arrays seperately
# (1 mark)
l = arr_data[:, 0]
l_var = arr_data[:, 1]
l_true = arr_data[:, 2]

```

```python
# step 3
# define the number of observation n
# (1 mark)
n = len(l)

# define the observation weight matrix P for LSE as identity matrix
# identity weight matrix simply means the observations will be unweighted in the following LSE
# (1 mark)
P = np.eye(n)

# define the range of function input(i.e., constants a), from start to end; and the numerical array of a
start = -1* np.pi
end = np.pi
a = np.linspace(start, end, n)


```

```python
# step 4
# use the matplotlib to plot the true and noisy observations together in one single figure.
# use scatter for noisy observation plotting
# for the true observation, label it as "f(a)"" and set color as 'k'
# for the noisy observation scatter plot, label it as "f(a) + e", set color as 'tab:blue', set marker as 'o' and set alpha as 0.3
# include proper axis labels, legend and title
# scale the range of x-axis to be -1 to 1
# keep this format for the rest of the plots in this lab assignment
# (2 marks)

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(14, 5))

ax.plot(a, l_true, label='f(a)', color='k')
ax.scatter(a, l, label='f(a) + e', color='tab:blue', marker= 'o', alpha= 0.3)
ax.set_xlabel('a')
ax.set_ylabel('f(a) + Noisy Observations')
ax.set_title('True Observations vs Noisy Observations')
ax.legend(loc='upper left')
ax.set_xlim(-1, 1)

ax.grid(True, which='both',ls='-',color='0.65')

fig.tight_layout()
```

```python
# step 5
# define a curve fitting function named fit_polynomial_curve
# the function should implement the linear least square estimation
# the function should include all important intermediate steps in LSE
# the input and output of the funciton are described in the comment below in red.
# please keep code clean and well-commented
# (7 marks)

def fit_polynomial_curve(a: np.ndarray, l: np.ndarray, P: np.ndarray, deg: int) -> tuple:
    '''
    Args:
        a:              function input (constants)
        l:              noisy observations
        P:              observation weight matrix
        deg:            degree of polynomial to fit to data
    Returns:
        x:              estimated polynomial coefficients
        cov_x:          uncertainty in estimated coefficients
        apv:            a posterier variance factor (goodness of fit)
        v:              observation residuals
    '''

    # Matrix for LSE multiplication
    #n=len(a) will calcualte the number of observation from the data
    #A = np.zeros will create a numpy array and fill it with zeros, and will be determined by the number of observations (n) and the deg +1 of the polynomial we are fitting for.
    # for i in range(n) and for j in range(deg +1), loops are made to iterate through data in the array,
    # and then are calcualted to raise the i value from the arary, and raises to the power of j
    n = len(a)
    A = np.zeros((n, deg + 1))
    for i in range(n):
        for j in range(deg + 1):
            A[i, j] = a[i]**j

    # This is the equation to solve for x in the weighted LSE equation
    # x = (A^T * P * A)^-1 * A^T * P * l

    # This is the system of normal equation for calculating least squares, we use @ in the function to solve for matrix multiplication questions.
    AtPA = A.T @ P @ A

    # This line calculates (A^T * P * A)^-1 by taking the inverse of the previous line (AtPA)
    AtPA_inv = np.linalg.inv(AtPA)

    # This line calculates A^T * P * l
    AtPl = A.T @ P @ l

    # Now that the equtions have been defined above, we cal calucalte the estimated coefficients x
    x = AtPA_inv @ AtPl

    # This line calcuates the observation residuals v = l- A @ x
    v = l - A @ x

    # This line calculates the varience in the measurement error, the posteriori variance factor (apv)
    # apv = (v^T * P * v) / (n - m), where n is number of observations, m is number of unknowns (coefficients), and m here is = to the deg + 1
    m = deg + 1
    apv = (v.T @ P @ v) / (n - m)

    # Lastly this line calculates the uncertainty in the estimated coefficients (cov_x)
    cov_x = apv * AtPA_inv

    return (x, cov_x, apv, v)
```

```python
# step 6
# call the fit_polynomial_curve function and fetch the results
# use polynomial degrees of: 2, 4, 6
# (1 marks)

x2= fit_polynomial_curve(a, l, P, 2)
x4= fit_polynomial_curve(a, l, P, 4)
x6= fit_polynomial_curve(a, l, P, 6)
```

```python
# step 7
# use the estimated coefficients x and constants array a, construct polynomial functions for each degree
# hint: use polyval function from numpy library (check input argument carefully)
# (1 marks)

Degree2 = np.polyval(x2[0], a)
Degree4 = np.polyval(x4[0], a)
Degree6 = np.polyval(x6[0], a)
```

```python
# step 8
# use the matplotlib to plot each polynomial function, noisy observations l, true observations l_true, together in one single figure
# for the polynomials of line, cubic and quintic, use color of orange, green and red respectively.
# (2 marks)

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(14, 5))

ax.plot(a, l_true, label='f(a)', color='k')
ax.scatter(a, l, label='f(a) + e', color='tab:blue', marker= 'o', alpha= 0.3)
ax.plot(a, Degree2, label='Polynomial Degree 2', color='orange')
ax.plot(a, Degree4, label='Polynomial Degree 4', color='green')
ax.plot(a, Degree6, label='Polynomial Degree 6', color='red')

ax.set_xlabel('a')
ax.set_ylabel('f(a) + Noisy Observations')
ax.set_title('True Observations, Noisy Observations, and Fitted Polynomials')
ax.legend(loc='upper left')




#I changed the Parameters from (-1, 1) to this to show a more clear graft viual, as before it did not look entirely correct.
ax.set_xlim(a.min(), a.max())


ax.grid(True, which='both',ls='-',color='0.65')

fig.tight_layout()
plt.show()
```
